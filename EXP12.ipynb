```python
#1. second datast
import pandas as pd
df = pd.read_csv("/content/Expt11_edp.csv")
df # or display(df)
```

```python
#2. Count students in each Grade.
# ...Find how many students got each grade.

print(df['Grade'].value_counts())
```

```python
#3. Count students of each Gender.
print(df['Gender'].value_counts())
```

```python
#4. Count students in each Department.
print(df['Department'].value_counts())
```

```python
#5. Percentage distribution of Grades.

print("Grade percentage")
print(df['Grade'].value_counts(normalize=True)*100)
```

```python
#6. Gender vs Grade Analysis.

print(pd.crosstab(df['Gender'],df['Grade']))
```

```python
#7. Department vs Gender Analysis.

print(pd.crosstab(df['Department'],df['Gender']))
```

```python
#8. Department vs Grade Analysis.

print(pd.crosstab(df['Department'],df['Grade']))
```

```python
#9. Department vs Grade Percentage Analysis.

print(pd.crosstab(df['Department'],df['Grade'],normalize='index')*100)
```

```python
#10. Grouping data
print(df.groupby('Department')['Grade'].value_counts())
```

```python
#11.
import pandas as pd

data = {
'Order_ID':['O1','O2','O3','O4','O5','O6','O7','O8','O9','O10'],

'Category':['Electronics','Clothing','Electronics','Grocery','Clothing',
            'Electronics','Grocery','Clothing','Electronics','Grocery'],

'Payment_Method':['UPI','Card','UPI','Cash','Card',
                  'UPI','Cash','Card','UPI','Cash'],

'Delivery_Type':['Express','Standard','Express','Standard','Express',
                 'Standard','Express','Standard','Express','Standard'],

'Customer_Type':['New','Returning','Returning','New','Returning',
                 'New','Returning','New','Returning','New']
}

df = pd.DataFrame(data)

display(df)
```

```python
#12. Frequency Count
#count how many orders in each category.

print(df['Category'].value_counts())
```

```python
#13. Percentge distribution
print(df['Category'].value_counts(normalize=True)*100)
```

```python
#14. Unique values.
#display all unique categories.

print(df['Category'].unique())
```

```python
#15. No. of Unique values.

print(df['Category'].nunique())
```

```python
#16. cross-tabulation.
#Category Vs payment method.

print(pd.crosstab(df['Category'],df['Payment_Method']))
```

```python
#17. Filtering Data
#Example: Orders from electronics category.

electonics_orders = df[df['Category']=='Electronics']
display(electonics_orders)
```

```python
#18. Sorting Data.
sort_df = df.sort_values(by='Category')
display(sort_df)
```

```python
#19.Gouping.
# Orders by category and payment method.
print(df.groupby('Category')['Payment_Method'].value_counts())
```
